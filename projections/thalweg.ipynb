## Estuarine analysis

```python
# mask the before and after temperature and salinities at the bdy
saltbase = np.where(base['vosaline'].mean('time').values[:,385:470,10]==0,np.nan,base['vosaline'].mean('time').values[:,385:470,10])
tempbase = np.where(base['votemper'].mean('time').values[:,385:470,10]==0,np.nan,base['votemper'].mean('time').values[:,385:470,10])
saltproj = np.where(bg_mean['vosaline'].values[:,385:470,10]==0,np.nan,bg_mean['vosaline'].values[:,385:470,10])
tempproj = np.where(bg_mean['votemper'].values[:,385:470,10]==0,np.nan,bg_mean['votemper'].values[:,385:470,10])
```

```python
fig, ax = plt.subplots(1,3,figsize=[8,4])

X, Y = np.meshgrid(bg_mean.gridY[385:470], bg_mean.depth)
ax[0].pcolormesh(X,Y,gsw.density.sigma0(saltbase,tempbase),
                 cmap=cm.deep,vmin=24.5,vmax=26.5)
ax[1].pcolormesh(X,Y,gsw.density.sigma0(saltproj,tempproj),
                 cmap=cm.deep,vmin=24.5,vmax=26.5)
ax[2].pcolormesh(X,Y,gsw.density.sigma0(saltproj,tempproj)-gsw.density.sigma0(saltbase,tempbase),
                 cmap=cm.balance,vmin=-0.4,vmax=+0.4)


for axs in ax:
    axs.set_ylim([300,0])
```

```python
# mean density of ocean (going in deeper cells to avoid mixing impact)
rho_base = np.nanmean(gsw.density.sigma0(saltbase,tempbase)[27:,:])+1000
rho_proj = np.nanmean(gsw.density.sigma0(saltproj,tempproj)[27:,:])+1000

print(f"202111 inflow density = {rho_base}")
print(f"projection inflow density = {rho_proj}")
```

```python
# attempting calculation from Guha and Lawrence 2013

#mean depth (m)
H = np.average(bg_mean.depth.values, weights=np.nansum(np.where(saltbase>0,1,np.nan),axis=1)*bg_mean.depth.values-np.append(0,bg_mean.depth.values[:-1]))
print(H)

#mean width (km)
B = np.average(np.nansum(np.where(saltbase>0,1,np.nan),axis=1), weights=bg_mean.depth.values-np.append(0,bg_mean.depth.values[:-1])) * 500
print(B)

# other variables we need 

Qr = 5000 # m3/s, mean annual eyeballed from Sutherland et al. 2011 and Allen et al. (2025)
# multiply by 2.2 to get the future amount

ut = 0.5 # tidal velocity in m/s estimated from Allen et al. (2025) figure 6


# coefficients
# beta = 7.7*(10**(-4)) # psu-1
rho_ref = 1025 # reference density.... range of 1000-1025, does not make a big difference
```

```python
g_red = 9.81*((rho_base-1000))/rho_ref
c = (g_red*H)**(1/2)

Fr = Qr/(B*H*c)
print(f"Fr = {Fr}")

Ft = (ut/c)*((B/H)**(1/2))
print(f"Ft = {Ft}")
```

```python
g_red = 9.81*((rho_proj-1000))/rho_ref
c = (g_red*H)**(1/2)

Fr = (2.2*Qr)/(B*H*c)
print(f"Fr = {Fr}")

Ft = (ut/c)*((B/H)**(1/2))
print(f"Ft = {Ft}")
```
