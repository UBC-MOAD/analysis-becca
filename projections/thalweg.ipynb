# Thalweg plots of best guess run mean results

```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
from cmocean import cm
import warnings

from salishsea_tools import visualisations as vis
import netCDF4 as nc

warnings.simplefilter('ignore')
```

```python
# bring in data
files = ['/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/All_2018main_primarytracers_forthalweg_20180101_20181231.nc',
         '/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/All_2019main_primarytracers_forthalweg_20190101_20191231.nc',
         '/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/All_2020main_primarytracers_forthalweg_20200101_20201231.nc',
         '/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/All_2021main_primarytracers_forthalweg_20210101_20211231.nc',
         '/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/All_2022main_primarytracers_forthalweg_20220101_20221231.nc']
bg18 = xr.open_dataset(files[0])
bg19 = xr.open_dataset(files[1])
bg20 = xr.open_dataset(files[2])
bg21 = xr.open_dataset(files[3])
bg22 = xr.open_dataset(files[4])
bg_mean = xr.open_mfdataset(files).mean('time')

base = xr.open_dataset('/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/forthalweg_physics202111_20180101_20221231.nc')
```

```python
# other grid files for thalweg plots
mesh = nc.Dataset('/home/sallen/MEOPAR/grid/mesh_mask201702.nc')
bathy = nc.Dataset("/home/sallen/MEOPAR/grid/bathymetry_202108.nc", mask_and_scale=False)
thalweg_in = '/home/sallen/MEOPAR/Tools/bathymetry/thalweg_working.txt'
```

```python
sal_range = np.arange(27,34,0.5)

fig, ax = plt.subplots(5,1,figsize=(8,8))

data_to_plot = bg18['vosaline'].values[0,:,:,:]
cbar = vis.contour_thalweg(ax[0], data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax[0].tick_params(axis='both', which='major', labelsize=12)

data_to_plot = bg19['vosaline'].values[0,:,:,:]
cbar = vis.contour_thalweg(ax[1], data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax[1].tick_params(axis='both', which='major', labelsize=12)

data_to_plot = bg20['vosaline'].values[0,:,:,:]
cbar = vis.contour_thalweg(ax[2], data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax[2].tick_params(axis='both', which='major', labelsize=12)

data_to_plot = bg21['vosaline'].values[0,:,:,:]
cbar = vis.contour_thalweg(ax[3], data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax[3].tick_params(axis='both', which='major', labelsize=12)

data_to_plot = bg22['vosaline'].values[0,:,:,:]
cbar = vis.contour_thalweg(ax[4], data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax[4].tick_params(axis='both', which='major', labelsize=12)

for axs in [ax[0],ax[1],ax[2],ax[3]]:
    axs.set_xticks(np.arange(0,701,100))
    axs.set_xticklabels([])
    axs.set_xlabel('')
```

```python
# 5 year mean 
fig, ax = plt.subplots(1,1,figsize=(8,2))

data_to_plot = bg_mean['vosaline'].values
cbar = vis.contour_thalweg(ax, data_to_plot, bathy, mesh, clevels=sal_range, cmap = cm.haline, thalweg_file=thalweg_in)
cbar.set_label('Salinity [g/kg]', fontsize = 12)
cbar.ax.tick_params(labelsize=12) 
ax.tick_params(axis='both', which='major', labelsize=12)
```
