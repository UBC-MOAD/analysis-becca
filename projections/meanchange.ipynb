```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
```

```python
colours = ['#CCBB44','#AA3377','#228833','#4477AA','#66CCEE','#EE6677']
scenarios = ['All','Salish','Local','North','Up','South']
```

```python
def theoretical(regionlist):
    theo = regionlist[1]+(regionlist[2]-regionlist[1])+(regionlist[3]-regionlist[1])+(regionlist[4]-regionlist[1])+(regionlist[5]-regionlist[1])
    return theo
```

## temperature

```python
# seasonal change at 50 m

fig, ax = plt.subplots(1,4,figsize=[9,3])

nsgchange = [np.mean(allnsg.votemper[:,23]-basensg.votemper[:,23]),
             np.mean(salishnsg.votemper[:,23]-basensg.votemper[:,23]),
             np.mean(localnsg.votemper[:,23]-basensg.votemper[:,23]),
             np.mean(northnsg.votemper[:,23]-basensg.votemper[:,23]),
             np.mean(upnsg.votemper[:,23]-basensg.votemper[:,23]),
             np.mean(southnsg.votemper[:,23]-basensg.votemper[:,23])] # need to add all and up later

csgchange = [np.mean(allcsg.votemper[:,23]-basecsg.votemper[:,23]),
             np.mean(salishcsg.votemper[:,23]-basecsg.votemper[:,23]),
             np.mean(localcsg.votemper[:,23]-basecsg.votemper[:,23]),
             np.mean(northcsg.votemper[:,23]-basecsg.votemper[:,23]),
             np.mean(upcsg.votemper[:,23]-basecsg.votemper[:,23]),
             np.mean(southcsg.votemper[:,23]-basecsg.votemper[:,23])]

jdfchange = [np.mean(alljdf.votemper[:,23]-basejdf.votemper[:,23]),
             np.mean(salishjdf.votemper[:,23]-basejdf.votemper[:,23]),
             np.mean(localjdf.votemper[:,23]-basejdf.votemper[:,23]),
             np.mean(northjdf.votemper[:,23]-basejdf.votemper[:,23]),
             np.mean(upjdf.votemper[:,23]-basejdf.votemper[:,23]),
             np.mean(southjdf.votemper[:,23]-basejdf.votemper[:,23])]

pugchange = [np.mean(allpug.votemper[:,23]-basepug.votemper[:,23]),
             np.mean(salishpug.votemper[:,23]-basepug.votemper[:,23]),
             np.mean(localpug.votemper[:,23]-basepug.votemper[:,23]),
             np.mean(northpug.votemper[:,23]-basepug.votemper[:,23]),
             np.mean(uppug.votemper[:,23]-basepug.votemper[:,23]),
             np.mean(southpug.votemper[:,23]-basepug.votemper[:,23])]

ax[0].bar(scenarios,nsgchange,color=colours)
ax[1].bar(scenarios,csgchange,color=colours)
ax[2].bar(scenarios,jdfchange,color=colours)
ax[3].bar(scenarios,pugchange,color=colours)

# add "theoretical linear change" to the all location
ax[0].bar(scenarios[0],theoretical(nsgchange),color='k',alpha=0.25)
ax[1].bar(scenarios[0],theoretical(csgchange),color='k',alpha=0.25)
ax[2].bar(scenarios[0],theoretical(jdfchange),color='k',alpha=0.25)
ax[3].bar(scenarios[0],theoretical(pugchange),color='k',alpha=0.25)

# formatting
for axx in ax:
    axx.set_ylim([-0.1,1.2])
    axx.set_xlim([-1,len(scenarios)])
    axx.set_xticks(np.arange(len(scenarios)))
    axx.set_xticklabels(scenarios,rotation=30)
    axx.hlines(xmin=-1,xmax=len(scenarios),y=-0,colors='k',linestyles='dashed',alpha=0.5)

ax[0].set_ylabel(r'Temperature Change ($^\circ$C)')
ax[0].set_title('N-SoG')
ax[1].set_title('C-SoG')
ax[2].set_title('JdF')
ax[3].set_title('Puget')


plt.tight_layout()
```
