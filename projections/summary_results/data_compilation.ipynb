# Taking a step further than reshapr
202111 data = taking spatial mean over region, putting tracers into one file <br>
projection data = spatial mean over region, aggregating 5 years of data and making climatology

```python
import xarray as xr
import numpy as np
```

```python
path = "/ocean/rbeutel/MOAD/analysis-becca/projections/summary_results/"
grid = xr.open_dataset("/home/sallen/MEOPAR/grid/bathymetry_202108.nc", mask_and_scale=False)
```

```python
def masking(regionname,data):

    if regionname == 'jdf':
        ymin,ymax = 310,360
        xmin,xmax = 50,100
    
    elif regionname == 'pug':
        ymin,ymax = 80,130
        xmin,xmax = 220,270

    elif regionname == 'nsg':
        ymin,ymax = 650,700
        xmin,xmax = 130,180

    elif regionname == 'csg' or regionname == 'cgs' :
        ymin,ymax = 460,510
        xmin,xmax = 240,290

    else:
        print("invalid region name")

    bathy = xr.DataArray(
        grid.Bathymetry[ymin:ymax, xmin:xmax].values,
        dims=("gridY", "gridX"),
        coords={
            "gridY": data.gridY,
            "gridX": data.gridX,
        },
    )

    # get cell range based on data depth
    z = data.depth.values
    z_top = np.empty_like(z, dtype=float)
    z_top[0] = 0.0 # first cell goes to surface
    z_top[1:] = 0.5 * (z[:-1] + z[1:]) # rest are halfway between cell centers

    z_top = xr.DataArray(
        z_top,
        dims="depth",
        coords={"depth": data.depth},
    )

    # A cell is wet if at least part of it lies above the seabed.
    wet = z_top < bathy

    # Broadcasts this over time, depth, gridY, gridX
    data_masked = data.where(wet)

    return data_masked
```

## 202111 results

```python
region= 'csg'

# load data into one xarray file
files = [f"{path}{region}_{j}202111_20180101_20221231.nc" for j in ['physics','chemistry','biology']]

data = xr.open_mfdataset(files)

# apply spatial mean over X and Y dimension
data_clim = (
    masking(region,data)
    .mean(dim=["gridY", "gridX"], skipna=True)
)

data_clim

# and dave
data_clim.to_netcdf(f'{path}{region}202111.nc')
```
